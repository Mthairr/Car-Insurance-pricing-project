# Inférence sur le modèle d'évaluation de risque de sinistre


```python
import pandas as pd
import pickle
import preparation
```

```python
initial_test_set = pd.read_csv("data/test.csv", sep=",")    
```

```python
with open("rf_frequence.pkl", "rb") as f:
    rf_final = pickle.load(f)

cols_final = list(rf_final.feature_names_in_)
print(cols_final)

p = rf_final.predict_proba(initial_test_set[cols_final])[:, 1]
```

```python
predictions = pd.DataFrame({
    "index": initial_test_set["index"],
    "proba": p,
})

display(predictions.head(10))
print(predictions.shape)
print(predictions["proba"].describe().round(4))
```
